# Functions

```python


def y_hat(x,a,b):
    return np.array((a*x)+b)

def mse(x,y,a,b):
    hat = y_hat(x,a,b)
    return np.sum((y-hat)**2)/hat.shape[0]

def mae(x,y,a,b):
    hat = y_hat(x,a,b)
    return np.sum(abs(y-hat))/hat.shape[0]

def rmse(x,a,b):
    return np.sqrt(mse(x,a,b))

def r2_score(x,y,a,b):
    hat = y_hat(x,a,b)
    return 1 - (np.sum((y-hat)**2) / np.sum((y-np.average(y))**2))

def normalization(x):
    z=np.zeros(x.shape[0])
    variance_x = (( np.sum( x - np.average(x)) **2) / x.shape[0]) ** 0.5
    z= (x - np.average(x))/variance_x
    return z
```
